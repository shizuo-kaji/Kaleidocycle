# Kaleidocycle Properties

Inspect the structural properties exposed by `Kaleidocycle` and `KaleidocycleAnimation`: size, orientation, closure, endpoint alignment, unit norms, and constant torsion.

```python
%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np

from kaleidocycle import Kaleidocycle, KaleidocycleAnimation, generate_animation, random_hinges
from kaleidocycle.constraints import alignment_residuals, closure_residual
```

## Single Configurations

```python
def summarize_kaleidocycle(kc: Kaleidocycle) -> dict[str, object]:
    return {
        "n": kc.n,
        "oriented": kc.oriented,
        "is_closed": kc.is_closed,
        "is_aligned": kc.is_aligned,
        "is_unit_norm": kc.is_unit_norm,
        "constant_torsion": kc.constant_torsion,
    }

kc_oriented = Kaleidocycle(8, oriented=True, seed=42)
kc_non_oriented = Kaleidocycle(7, oriented=False, seed=42)
kc_random = Kaleidocycle(hinges=random_hinges(6, seed=123).as_array())

for name, kc in {
    "oriented n=8": kc_oriented,
    "non-oriented n=7": kc_non_oriented,
    "random hinges n=6": kc_random,
}.items():
    print(f"
{name}")
    for key, value in summarize_kaleidocycle(kc).items():
        print(f"  {key}: {value}")
```

## Residuals Behind the Convenience Properties

```python
kc = Kaleidocycle(6, oriented=True, seed=42)
closure = closure_residual(kc.hinges, slide=0.0)
alignment = alignment_residuals(kc.hinges, oriented=kc.oriented)

print(f"closure residual norm: {np.linalg.norm(closure):.3e}")
print(f"alignment residual: {alignment:.3e}")
print(f"unit norm range: {np.linalg.norm(kc.hinges, axis=1).min():.6f} to {np.linalg.norm(kc.hinges, axis=1).max():.6f}")
```

## Animation-Wide Properties

For animations, these properties are checked over all frames.

```python
frames = generate_animation(
    kc.hinges,
    num_frames=15,
    step_size=0.02,
    rule="sine-Gordon",
    oriented=kc.oriented,
)
anim = KaleidocycleAnimation(frames=frames, evolution_rule="sine-Gordon")

print(f"frames: {anim.n_frames}")
print(f"n: {anim.n}")
print(f"oriented: {anim.oriented}")
print(f"all frames closed: {anim.is_closed}")
print(f"all frames aligned: {anim.is_aligned}")
print(f"all frames unit norm: {anim.is_unit_norm}")
print(f"constant torsion: {anim.constant_torsion}")
```
